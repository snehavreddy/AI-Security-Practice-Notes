```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

from langchain.chat_models import init_chat_model

model=init_chat_model("groq:openai/gpt-oss-20b")
```

#### Streaming

```python
# Normal response > Generates o/p only once entire response is generated
model.invoke("Write me a 200 words paragraph on Ariticial Intelligence")
```

```python
# Using stream()
for chunk in model.stream("Write me a 200 words paragraph on Ariticial Intelligence"):
    print(chunk.text, end="", flush=True)
    # print(chunk.text, end="|", flush=True) # Just to demo chunks
```
